```python
from sentence_transformers import SentenceTransformer
from tqdm.auto import tqdm

from llm_zoomcamp.ingest import ingest_faq_data

model = SentenceTransformer("all-MiniLM-L6-v2")

documents = ingest_faq_data()
texts = [doc["question"] + " " + doc["answer"] for doc in documents]

batch_size = 50
vectors = []

for i in tqdm(range(0, len(texts), batch_size)):
    batch = texts[i : i + batch_size]
    batch_vectors = model.encode(batch)
    vectors.extend(batch_vectors)
```

```python
import psycopg

conn = psycopg.connect("postgresql://postgres:postgres@localhost:5433/postgres")

conn.execute("Create extension if not exists vector")
```

```python
conn.execute("""
    DROP TABLE IF EXISTS documents
""")

conn.execute("""
    CREATE TABLE documents (
        id SERIAL PRIMARY KEY,
        course TEXT,
        section TEXT,
        question TEXT,
        answer TEXT,
        embedding vector(384)
    )
""")
```

```python
def vec_to_str(vector):
    return "[" + ",".join(str(x) for x in vector) + "]"


for doc, vect in tqdm(zip(documents, vectors), total=len(documents)):
    conn.execute(
        """
        insert into documents (course,section,question,answer,embedding)
        values(%s,%s,%s,%s,%s::vector)
        """,
        (doc["course"], doc["section"], doc["question"], doc["answer"], vec_to_str(vect)),
    )
conn.commit()
```

```python
query = "I just discovered the course. Can i still join it?"
query_vector = model.encode(query)
query_str = vec_to_str(query_vector)
```

```python
results = conn.execute(
    """
    SELECT course,question,answer, 1 - (embedding <=> %s::vector) as similarity
    FROM documents
    ORDER BY embedding <=> %s::vector
    LIMIT 5
    """,
    (query_str, query_str),
).fetchall()

for row in results:
    print(f"[{row[0]}] {row[1]} (similarity: {row[3]:.4f})")
```

```python
def pgvector_search(query, course="llm-zoomcamp", num_results=5):
    query_vector = model.encode(query)
    query_str = vec_to_str(query_vector)

    rows = conn.execute(
        """
        SELECT Course, section,question, answer
        from documents
        Where course = %s
        Order by embedding <=> %s::vector
        Limit %s
        """,
        (course, query_str, num_results),
    ).fetchall()

    return [{"course": r[0], "section": r[1], "question": r[2], "answer": r[3]} for r in rows]
```

```python
from llm_zoomcamp.rag_helper import RAGBase


class RAGPgVector(RAGBase):
    def __init__(self, embedder, conn, **kwargs):
        super().__init__(index=None, **kwargs)
        self.embedder = embedder
        self.conn = conn

    def search(self, query, num_results=5):
        query_vector = self.embedder.encode(query)
        query_str = vec_to_str(query_vector)

        rows = self.conn.execute(
            """
            SELECT Course, section,question, answer
            from documents
            Where course = %s
            Order by embedding <=> %s::vector
            Limit %s
            """,
            (self.course, query_str, num_results),
        ).fetchall()

        return [{"course": r[0], "section": r[1], "question": r[2], "answer": r[3]} for r in rows]
```

```python
from dotenv import load_dotenv
from google import genai
from google.genai import types

load_dotenv()

SYSTEM_INSTRUCTION = """
Your task is to answer the questions from the course participants based on the given context
    Use the context to find relevant information and provide accurate answer to the user
    if there is no answer in the context do not make up things just tell the user that you don't know
"""

client = genai.Client()
config = types.GenerateContentConfig(system_instruction=SYSTEM_INSTRUCTION)

vassistant = RAGPgVector(
    llm_client=client, config=config, model="gemini-3.6-flash", embedder=model, conn=conn
)
```
